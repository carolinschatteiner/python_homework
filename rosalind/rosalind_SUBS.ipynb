# Problem - SUBS

- important: Rosalind "location" is human readable, do index+1
  
- Given: Two DNA strings s and t (each of length at most 1 kbp).

- Return: All locations of t as a substring of s.

```python
for i in range(len(s)-len(t)): # - len(t) because we want to stop at the end of the whole string minus our substring
    print(s[i:i+len(t)])
```

```python
steps = len(s) - len(t) + 1
sub_length = len(t)
substrings = []
positions = []
for i in range(steps): 
    if s[i:i+sub_length] == t:
        substrings.append(s[i:i+sub_length])
        positions.append(i+1)

print(positions)
        
```
